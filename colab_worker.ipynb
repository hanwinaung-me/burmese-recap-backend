ဒီ Notebook က Video Processing ကို Colab မှာ လုပ်ပေးပါတယ်။ Render API က Job စာရင်းနဲ့ Progress ကိုသာ ထိန်းပါတယ်။

**အရင်ပြင်ဆင်ရန်**
1. Project ZIP ထဲက `colab_worker.py` ဖိုင်ကို ဖုန်းထဲမှာ Extract လုပ်ထားပါ။
2. Render မှာ `API_BASE_URL` (ဥပမာ `https://your-api.onrender.com`) နဲ့ `WORKER_TOKEN` ရှိရပါမယ်။
3. Google Drive ထဲမှာ ကိုယ့် Video ကို Upload လုပ်ပြီး File ID ကို Lovable App မှာ ထည့်ပါ။
4. Gemini API Key ကို Google AI Studio မှာ ဖန်တီးပါ။

Colab Free ရဲ့ Runtime/Resource ကန့်သတ်ချက်တွေကြောင့် အချိန်ကြာတဲ့ Job အတွက် Runtime ပြတ်နိုင်ပါတယ်။ Notebook ကို run နေစေပြီး Tab ကို မပိတ်ထားပါနှင့်။ Runtime ပြတ်သွားရင် ပြန် run လုပ်နိုင်ပါတယ်။
